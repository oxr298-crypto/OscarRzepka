```python
import random
#in-class assignment 2/11/26
start = input("Do you want to play a game? (y/n) ")
while start.lower() == "y":
    computer_move = random.choice(['rock', 'paper', 'scissors'])
    player_move = input("your move: ").lower()
    print(f"Computer Move: {computer_move}")

    if player_move == computer_move:
        print("It's a tie!")
    elif player_move == 'rock':
        if computer_move == 'scissors':
            print("You win!")
        else:
            print("Computer wins!")
    elif player_move == 'paper':
        if computer_move == 'rock':
            print("You win!")
        else:
            print("Computer wins!")
    elif player_move == 'scissors':
        if computer_move == 'paper':
            print("You win!")
        else:
            print("Computer wins!")
    else:
        print("Invalid move. Please choose rock, paper, or scissors.")

    start = input("Do you want to play again? (y/n) ")
print("Game over!")
```

```python
def getValues(n):
  items=[]

  for i in range (n):
    items.append(float(input("enter a value: ")))

  return items

def calcAverage(alist):
  assert isinstance(alist, list), "'alist' must be a list"
  try:
    return sum(alist[2:]) /len(alist[2:])
  except ZeroDivisionError:
    print("the list 'alist' must have at least 3 items")
    return None
  except:
    raise ValueError("the length of list 'alist' cannot be < 2")

def main():
  n= int(input("enter the # of items:  "))
  alist = getValues(n)
  average = calcAverage(alist)
  if average is not None:
    print(f"The average of the list (excluding the first two items) is: {average}")

main()
```

```python
class Employee:
  def _init_(self, name, wage, hours):
    self.name = name
    self._wage = wage
    self.hours = hours

def get_wage(self):
  return self._wage

def _setwage (self, neWage):
  if neWage <= 0:
    print("negative wage values not allowed")
  else:
    self._wage = neWage
def calculate_pay(self):
  return self._wage * self.hours

def display_info(self):
  print(f"Employee: {self.name} hourly wages: {self._wage}")
```

```python
class Greeting:
  def __init__(self, first, last):
    self.first = first
    self.last = last

  def display(self, formal= False):
    if formal:
      print(f"{self.last}, {self.first}")
    else:
      print(self.first)

  def __str__(self):
    return "Instance of class greeting"

name = Greeting("John", "Doe")
print(name)
name.display()
name.display(formal=True)
```
